```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet", "Monitor", "Keyboard"],
    "Category": ["Electronics", "Electronics", "Electronics", "Electronics", "Accessories"],
    "Sales": [50000, 35000, 20000, 15000, 5000],
    "Quantity": [5, 7, 4, 3, 10]
}

df = pd.DataFrame(data)

df.to_csv("w7d5_sales.csv", index=False)

print("CSV file created successfully!")
print(df)
```

```python
import pandas as pd

df = pd.read_csv("w7d5_sales.csv")

print("CSV loaded successfully!")
print(df)
```

```python
print("Data stored in memory successfully!")
print("Rows:", len(df))
print("Columns:", list(df.columns))
print("\nDataset:")
print(df)
```

```python
def data_analyst_query(question):
    question = question.lower()

    if "total sales" in question:
        return f"Total Sales: ₹{df['Sales'].sum():,}"

    elif "highest sales" in question:
        row = df.loc[df["Sales"].idxmax()]
        return f"Highest Sales: {row['Product']} - ₹{row['Sales']:,}"

    elif "lowest sales" in question:
        row = df.loc[df["Sales"].idxmin()]
        return f"Lowest Sales: {row['Product']} - ₹{row['Sales']:,}"

    elif "total quantity" in question:
        return f"Total Quantity: {df['Quantity'].sum()}"

    elif "average sales" in question:
        return f"Average Sales: ₹{df['Sales'].mean():,.2f}"

    else:
        return "Question not supported."


print(data_analyst_query("What is the total sales?"))
```

```python
import json

chart_data = {
    "x": df["Product"].tolist(),
    "y": df["Sales"].tolist(),
    "type": "bar"
}

plotly_json = json.dumps(chart_data)

print("Chart data generated successfully!")
print(plotly_json)
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x="Product",
    y="Sales",
    title="Product-wise Sales"
)

fig.show()
```

```python
questions = [
    "What is the total sales?",
    "What is the highest sales?",
    "What is the lowest sales?",
    "What is the total quantity?",
    "What is the average sales?"
]

for question in questions:
    print("Question:", question)
    print("Answer:", data_analyst_query(question))
    print("-" * 40)
```

```python
def analyst_response(question):
    answer = data_analyst_query(question)

    response = {
        "question": question,
        "answer": answer,
        "chart_data": chart_data
    }

    return response


result = analyst_response("What is the total sales?")

print(result)
```

```python
question = "What is the highest sales?"

result = analyst_response(question)

print("QUESTION:")
print(result["question"])

print("\nANSWER:")
print(result["answer"])

print("\nCHART DATA:")
print(result["chart_data"])
```
