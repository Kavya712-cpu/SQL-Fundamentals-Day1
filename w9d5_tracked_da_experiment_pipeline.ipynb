```python
import mlflow
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample Data Analyst sales data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([100, 200, 300, 400, 500])

# Create MLflow experiment
mlflow.set_experiment("W9D5_Tracked_DA_Experiment_Pipeline")

# Start experiment run
with mlflow.start_run(run_name="DA_Sales_Experiment") as run:

    # Train model
    model = LinearRegression()
    model.fit(X, y)

    # Make predictions
    predictions = model.predict(X)

    # Calculate metric
    mse = mean_squared_error(y, predictions)

    # Log experiment information
    mlflow.log_param("model_type", "Linear Regression")
    mlflow.log_param("data_points", len(X))
    mlflow.log_metric("mse", mse)

    print("Tracked DA Experiment Pipeline completed successfully")
    print("Model:", "Linear Regression")
    print("Data Points:", len(X))
    print("MSE:", mse)
    print("Run ID:", run.info.run_id)
```
