```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Create sample monthly sales data
dates = pd.date_range(start="2024-01-01", periods=24, freq="MS")

sales = [
    120, 135, 128, 145, 160, 155,
    170, 180, 175, 190, 205, 215,
    210, 225, 218, 235, 250, 245,
    260, 270, 265, 280, 295, 305
]

df = pd.DataFrame({
    "Date": dates,
    "Sales": sales
})

df.set_index("Date", inplace=True)

# Display data
print("Time Series Data:")
display(df)

# Plot sales trend
plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# ARIMA model
model = ARIMA(df["Sales"], order=(1, 1, 1))
model_fit = model.fit()

# Forecast next 6 months
forecast = model_fit.forecast(steps=6)

forecast_dates = pd.date_range(
    start=df.index[-1] + pd.DateOffset(months=1),
    periods=6,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Date": forecast_dates,
    "Forecasted_Sales": forecast.values
})

print("ARIMA Forecast:")
display(forecast_df)

print("Time Series Analysis completed successfully.")
```

# W10D1 — Time Series Analysis: Trends, Seasonality, ARIMA

## Objective
To analyze monthly sales data and forecast future sales using ARIMA.

## Work Done
- Created sample monthly sales data for 24 months.
- Plotted the sales values to understand the trend.
- Applied the ARIMA model with order (1,1,1).
- Forecasted sales for the next 6 months.
- Displayed the forecast results.

## Key Concepts
- **Trend:** The overall increase or decrease in values over time.
- **Seasonality:** Repeating patterns that occur at regular time intervals.
- **ARIMA:** A time-series forecasting model used to predict future values from past data.

## Result
The ARIMA model generated a 6-month sales forecast successfully.
