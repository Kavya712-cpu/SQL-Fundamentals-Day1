```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("teen_phone_addiction_dataset.csv")

df.head()
```

```python
column = "Academic_Performance"

print("Mean:", df[column].mean())
print("Median:", df[column].median())
print("Mode:", df[column].mode()[0])
print("Variance:", df[column].var())
print("Standard Deviation:", df[column].std())
```

```python
plt.hist(df["Academic_Performance"], bins=10)

plt.title("Frequency Distribution of Academic Performance")
plt.xlabel("Academic Performance")
plt.ylabel("Frequency")

plt.show()
```

```python
print("25th Percentile (Q1):", df["Academic_Performance"].quantile(0.25))
print("50th Percentile (Median):", df["Academic_Performance"].quantile(0.50))
print("75th Percentile (Q3):", df["Academic_Performance"].quantile(0.75))
```

```python
df.boxplot(column="Academic_Performance", by="Gender")

plt.title("Academic Performance by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Academic Performance")

plt.show()
```
