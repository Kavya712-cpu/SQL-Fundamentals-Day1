```python
import mlflow
import numpy as np

# Create sample results for two model versions
model_versions = {
    "Version 1": {"accuracy": 0.82, "mae": 0.18},
    "Version 2": {"accuracy": 0.89, "mae": 0.11}
}

# Create MLflow experiment
mlflow.set_experiment("W9D4_Model_Version_Comparison")

# Log Version 1
with mlflow.start_run(run_name="Model_Version_1") as run1:
    mlflow.log_param("model_version", "Version 1")
    mlflow.log_metric("accuracy", model_versions["Version 1"]["accuracy"])
    mlflow.log_metric("mae", model_versions["Version 1"]["mae"])
    
    print("Version 1 logged")
    print("Run ID:", run1.info.run_id)

# Log Version 2
with mlflow.start_run(run_name="Model_Version_2") as run2:
    mlflow.log_param("model_version", "Version 2")
    mlflow.log_metric("accuracy", model_versions["Version 2"]["accuracy"])
    mlflow.log_metric("mae", model_versions["Version 2"]["mae"])
    
    print("Version 2 logged")
    print("Run ID:", run2.info.run_id)

# Compare the versions
if model_versions["Version 2"]["accuracy"] > model_versions["Version 1"]["accuracy"]:
    better_version = "Version 2"
else:
    better_version = "Version 1"

print("\nModel Version Comparison")
print("------------------------")
print("Version 1 Accuracy:", model_versions["Version 1"]["accuracy"])
print("Version 2 Accuracy:", model_versions["Version 2"]["accuracy"])
print("Version 1 MAE:", model_versions["Version 1"]["mae"])
print("Version 2 MAE:", model_versions["Version 2"]["mae"])
print("Better Accuracy:", better_version)
```
