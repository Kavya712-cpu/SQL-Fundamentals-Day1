```python
import mlflow
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Set experiment
mlflow.set_experiment("W9D2_Model_Registry")

with mlflow.start_run() as run:
    model = LinearRegression()
    model.fit(X, y)

    # Log model
    model_info = mlflow.sklearn.log_model(
        model,
        "linear_regression_model"
    )

    print("Model trained successfully")
    print("Run ID:", run.info.run_id)
    print("Model URI:", model_info.model_uri)

    # Register model
    registered_model = mlflow.register_model(
        model_info.model_uri,
        "DA_Linear_Regression"
    )

    print("Model registered successfully")
    print("Model Name:", registered_model.name)
    print("Model Version:", registered_model.version)
```

# W9D2 - MLflow Model Registry for Analyst Models

## Objective
To register and version an analyst machine learning model using MLflow Model Registry.

## Work Done
- Created sample data.
- Trained a Linear Regression model.
- Logged the model using MLflow.
- Registered the model as `DA_Linear_Regression`.
- Created Model Version 1.

## Result
The model was successfully registered in MLflow Model Registry.

Model Name: DA_Linear_Regression  
Model Version: 1
