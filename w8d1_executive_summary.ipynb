```python
import pandas as pd

df = pd.read_csv("startup_funding.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())
```

```python
print("Dataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nBasic Statistics:")
print(df.describe(include="all"))
```

```python
# Clean the funding amount column

df["Amount in USD"] = (
    df["Amount in USD"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df["Amount in USD"] = pd.to_numeric(
    df["Amount in USD"],
    errors="coerce"
)

# Generate 5 insights

print("5 DATA INSIGHTS")
print("=" * 40)

print("1. Total startups:", len(df))

print("2. Total funding: $", round(df["Amount in USD"].sum(), 2))

print("3. Average funding: $", round(df["Amount in USD"].mean(), 2))

print("4. Highest funding: $", round(df["Amount in USD"].max(), 2))

print("5. Lowest funding: $", round(df["Amount in USD"].min(), 2))
```

```python
insights = [
    f"Total number of startups: {len(df)}",
    f"Total funding raised: ${df['Amount in USD'].sum():,.2f}",
    f"Average funding per startup: ${df['Amount in USD'].mean():,.2f}",
    f"Highest funding received by a startup: ${df['Amount in USD'].max():,.2f}",
    f"Lowest recorded funding amount: ${df['Amount in USD'].min():,.2f}"
]

print("INSIGHTS FOR EXECUTIVE SUMMARY")
print("=" * 50)

for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")
```

```python
prompt = f"""
You are a professional Data Analyst.

Create an executive summary of approximately 200 words based on the
following 5 data insights from a startup funding dataset.

Data Insights:
1. {insights[0]}
2. {insights[1]}
3. {insights[2]}
4. {insights[3]}
5. {insights[4]}

The summary should:
- Be clear and professional
- Highlight the important findings
- Explain what the findings mean for decision makers
- Avoid making up information that is not present in the data
- End with useful business recommendations
"""

print(prompt)
```

```python
executive_summary = """
The startup funding dataset provides an overview of funding activity
across multiple startups. The analysis identifies the total number of
startups, overall funding raised, average funding per startup, and the
highest and lowest funding amounts recorded.

The funding distribution shows that funding levels vary significantly
between startups, indicating differences in their investment and growth
requirements. The startups receiving higher funding may require greater
capital to support expansion, technology development, market growth, or
operations.

From a business perspective, the funding information can help decision
makers understand investment patterns and identify startups receiving
significant financial support. The results can also be used to compare
funding levels and identify areas requiring further investigation.

It is recommended to analyze funding by industry, location, year, and
investment type to understand the factors influencing funding levels.
Further analysis can help investors and business teams make better
investment and strategic decisions.
"""

print("EXECUTIVE SUMMARY")
print("=" * 60)
print(executive_summary)
```

```python
from pptx import Presentation
from pptx.util import Inches, Pt

# Create presentation
prs = Presentation()

# ---------- Slide 1: Title ----------
slide = prs.slides.add_slide(prs.slide_layouts[0])
slide.shapes.title.text = "Startup Funding Analysis"
slide.placeholders[1].text = "W8D1 – Executive Summary Generation"

# ---------- Slides 2–6: Five Insights ----------
for i, insight in enumerate(insights, 1):
    slide = prs.slides.add_slide(prs.slide_layouts[1])
    slide.shapes.title.text = f"Insight {i}"
    slide.placeholders[1].text = insight

# ---------- Slide 7: Recommendations ----------
slide = prs.slides.add_slide(prs.slide_layouts[1])
slide.shapes.title.text = "Recommendations"

slide.placeholders[1].text = (
    "• Analyze funding by industry, location and year.\n"
    "• Compare funding levels across startups.\n"
    "• Investigate factors affecting high and low funding.\n"
    "• Use funding trends to support investment decisions.\n"
    "• Perform deeper analysis for better business planning."
)

# Save presentation
ppt_file = "W8D1_Startup_Funding_Executive_Summary.pptx"
prs.save(ppt_file)

print("PowerPoint created successfully!")
print("File:", ppt_file)
print("Total slides:", len(prs.slides))
```
