```python
from vanna.integrations.sqlite import SqliteRunner

runner = SqliteRunner("w7d4_students.db")

print("SQLite runner created successfully!")
```

```python
result = runner.run_sql(
    "SELECT * FROM students LIMIT 5",
    {}
)

print(result)
```

```python
import inspect

print(inspect.signature(runner.run_sql))
print(inspect.getsource(runner.run_sql))
```

```python
from vanna.capabilities.sql_runner.models import RunSqlToolArgs
```

```python
args = RunSqlToolArgs(
    sql="SELECT * FROM students LIMIT 5"
)
```

```python
result = await runner.run_sql(args, None)
print(result)
```
