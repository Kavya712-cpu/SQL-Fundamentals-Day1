```python
import pandas as pd

df = pd.read_csv("startup_funding.csv")

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
print("W8D2 - BASIC EDA")
print("=" * 50)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
print("W8D2 - FIVE DATA INSIGHTS")
print("=" * 50)

# 1. Total number of startups
total_startups = len(df)

# 2. Missing values
total_missing = df.isnull().sum().sum()

# 3. Dataset size
rows, columns = df.shape

# 4. Numeric columns
numeric_columns = df.select_dtypes(include="number").columns.tolist()

# 5. Categorical columns
categorical_columns = df.select_dtypes(include="object").columns.tolist()

insights = [
    f"The dataset contains {total_startups} startup records.",
    f"The dataset contains {rows} rows and {columns} columns.",
    f"There are {total_missing} missing values across the dataset.",
    f"The dataset contains {len(numeric_columns)} numerical columns for quantitative analysis.",
    f"The dataset contains {len(categorical_columns)} categorical columns for category-based analysis."
]

for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")
```

```python
print("W8D2 - BUSINESS RECOMMENDATION PROMPT")
print("=" * 50)

recommendation_prompt = """
You are a Business Data Analyst.

Analyze the following five data insights from a startup funding dataset.

Your task is to:
1. Explain what the insights mean for a business.
2. Identify possible business opportunities or risks.
3. Give practical recommendations for decision makers.
4. Keep the recommendations simple and actionable.

Data Insights:
"""

for i, insight in enumerate(insights, 1):
    recommendation_prompt += f"\n{i}. {insight}"

recommendation_prompt += """

Provide 5 clear business recommendations based on these insights.
Use simple professional language.
"""

print(recommendation_prompt)
```

```python
print("W8D2 - BUSINESS RECOMMENDATIONS")
print("=" * 50)

recommendations = [
    "1. Focus on startups with strong funding potential and clear growth opportunities.",
    "2. Compare funding levels across different startup categories to identify promising sectors.",
    "3. Investigate startups with unusually high funding to understand the factors behind investor interest.",
    "4. Improve data quality by addressing missing values before making important business decisions.",
    "5. Regularly analyze funding trends to support better investment and strategic planning."
]

for recommendation in recommendations:
    print(recommendation)
```

```python
from pptx import Presentation
from pptx.util import Inches, Pt

prs = Presentation()

# Title slide
slide = prs.slides.add_slide(prs.slide_layouts[0])
slide.shapes.title.text = "Startup Funding Analysis"
slide.placeholders[1].text = "Business Recommendations — W8D2"

# 5 Insight slides
for i, insight in enumerate(insights, 1):
    slide = prs.slides.add_slide(prs.slide_layouts[1])
    slide.shapes.title.text = f"Data Insight {i}"
    slide.placeholders[1].text = insight

# Recommendation slide
slide = prs.slides.add_slide(prs.slide_layouts[1])
slide.shapes.title.text = "Business Recommendations"

slide.placeholders[1].text = "\n".join(recommendations)

# Save PPT
ppt_file = "W8D2_Startup_Funding_Business_Recommendations.pptx"
prs.save(ppt_file)

print("PowerPoint created successfully!")
print("File:", ppt_file)
print("Total slides:", len(prs.slides))
```

```python
print("W8D2 - COMPLETE WORKFLOW TEST")
print("=" * 50)

print("1. CSV loaded successfully       ✓")
print("2. Pandas EDA completed          ✓")
print("3. 5 data insights generated    ✓")
print("4. AI recommendation prompt     ✓")
print("5. 5 business recommendations   ✓")
print("6. PowerPoint exported          ✓")

print("\nComplete W8D2 workflow tested successfully!")
```
