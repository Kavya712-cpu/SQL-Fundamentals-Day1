```python
import pandas as pd

df = pd.read_csv("startup_funding.csv")

print(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
from pandasai import SmartDataframe

sdf = SmartDataframe(df)

print("PandasAI setup completed!")
```

```python
# 10 Natural Language questions prepared for PandasAI

questions = [
    "How many startups are there?",
    "Which city has the most startups?",
    "Which industry has the most startups?",
    "What is the most common investment type?",
    "Which startup received the highest investment?",
    "How many startups are from Bengaluru?",
    "How many startups are from Mumbai?",
    "Which investors invested in the most startups?",
    "What is the average investment amount?",
    "Which year had the most startup entries?"
]

for i, question in enumerate(questions, 1):
    print(f"{i}. {question}")
```

```python
print("1. Total startups:", df["Startup Name"].nunique())

print("2. Top cities:")
print(df["City  Location"].value_counts().head())

print("3. Top industries:")
print(df["Industry Vertical"].value_counts().head())

print("4. Investment types:")
print(df["InvestmentnType"].value_counts().head())

print("5. Dataset shape:", df.shape)
```

```python
comparison = {
    "Question 1": "How many startups are there?",
    "Question 2": "Which city has the most startups?",
    "Question 3": "Which industry has the most startups?"
}

for question, value in comparison.items():
    print(question, ":", value)
```

```python
def ask_pandasai(question):
    try:
        result = sdf.chat(question)
        return result
    except Exception as e:
        print("PandasAI query failed.")
        print("Reason:", e)
        return None

ask_pandasai("How many startups are there?")
```
