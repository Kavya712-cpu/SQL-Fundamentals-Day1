```python
import polars as pl
import pandas as pd
import time

print("Libraries imported successfully!")
```

```python
pl_df = pl.read_csv("teen_phone_addiction_dataset.csv")

print(pl_df.head())
print(pl_df.shape)
```

```python
print(pl_df.schema)
```

```python
pl_df.select([
    "Daily_Usage_Hours",
    "Academic_Performance"
]).head()
```

```python
pl_df.filter(
    pl.col("Daily_Usage_Hours") > 5
)
```

```python
pl_df.sort(
    "Daily_Usage_Hours",
    descending=True
).head(10)
```

```python
pl_df.group_by("Gender").agg(
    pl.col("Daily_Usage_Hours").mean().alias("Average_Usage")
)
```

```python
pl_df.select([
    pl.col("Daily_Usage_Hours").mean().alias("Average Usage"),
    pl.col("Daily_Usage_Hours").max().alias("Maximum Usage"),
    pl.col("Daily_Usage_Hours").min().alias("Minimum Usage")
])
```

```python
import time

# Pandas benchmark
pandas_df = pd.read_csv("teen_phone_addiction_dataset.csv")

start = time.time()
pandas_df["Daily_Usage_Hours"].mean()
pandas_time = time.time() - start

# Polars benchmark
start = time.time()
pl_df.select(pl.col("Daily_Usage_Hours").mean())
polars_time = time.time() - start

print(f"Pandas Time: {pandas_time:.6f} seconds")
print(f"Polars Time: {polars_time:.6f} seconds")
```
