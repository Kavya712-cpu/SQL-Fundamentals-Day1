```python
import pandas as pd
from pptx import Presentation
from pptx.util import Inches, Pt

print("W8D3 - PPT Report Export")
print("Libraries imported successfully!")
```

```python
# Load startup funding dataset

df = pd.read_csv("w7d5_sales.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Basic EDA

print("DATASET SUMMARY")
print("=" * 40)

print("Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
# Create 5 insights from the dataset

print("5 DATA INSIGHTS")
print("=" * 40)

print("1. The dataset contains", len(df), "records.")
print("2. The dataset contains", len(df.columns), "columns.")
print("3. Missing values were checked across all columns.")
print("4. The dataset can be used to identify important business patterns.")
print("5. These insights can be presented in a management-friendly PPT report.")
```

```python
# Executive summary for PPT

executive_summary = """
The analysis of the startup funding dataset provides a useful overview
of funding patterns and business activity. The data contains multiple
records and variables that can help identify important business trends.
Missing values and data quality were reviewed before preparing the report.
The analysis can help compare startups, understand funding patterns,
and support better business planning and investment decisions.
"""

print("EXECUTIVE SUMMARY")
print("=" * 40)
print(executive_summary)
```

```python
# Create PowerPoint report

prs = Presentation()

# Title slide
slide = prs.slides.add_slide(prs.slide_layouts[0])
slide.shapes.title.text = "Startup Funding Analysis"
slide.placeholders[1].text = "W8D3 - PPT Report Export"

# Executive Summary slide
slide = prs.slides.add_slide(prs.slide_layouts[1])
slide.shapes.title.text = "Executive Summary"
slide.placeholders[1].text = executive_summary

# 5 Insight slides
for i, insight in enumerate([
    "The dataset contains multiple startup funding records.",
    "The dataset includes several business-related variables.",
    "Missing values were checked during data preparation.",
    "The data can be used to identify funding and business patterns.",
    "The insights can support business planning and investment decisions."
], 1):
    slide = prs.slides.add_slide(prs.slide_layouts[1])
    slide.shapes.title.text = f"Insight {i}"
    slide.placeholders[1].text = insight

# Recommendation slide
slide = prs.slides.add_slide(prs.slide_layouts[1])
slide.shapes.title.text = "Business Recommendations"
slide.placeholders[1].text = (
    "• Monitor startup funding trends regularly.\n"
    "• Compare funding patterns across startups.\n"
    "• Improve data quality before decision-making.\n"
    "• Use insights to support investment planning."
)

# Save PPT
output_file = "W8D3_Startup_Funding_PPT_Report.pptx"
prs.save(output_file)

print("PPT created successfully!")
print("File:", output_file)
```
