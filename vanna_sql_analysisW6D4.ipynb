```python
import vanna
from vanna.local import LocalContext_OpenAI

vn = LocalContext_OpenAI(config={
    "api_key": "dummy"
})

print(vn)
```

```python
vn.train(
    question="Show all students.",
    sql="SELECT * FROM students;"
)

vn.train(
    question="Show male students.",
    sql="SELECT * FROM students WHERE Gender='Male';"
)

vn.train(
    question="Show female students.",
    sql="SELECT * FROM students WHERE Gender='Female';"
)

vn.train(
    question="Show students using phone more than 6 hours.",
    sql="SELECT * FROM students WHERE Daily_Usage_Hours > 6;"
)

vn.train(
    question="Show top 5 academic performers.",
    sql="SELECT * FROM students ORDER BY Academic_Performance DESC LIMIT 5;"
)
```
