```python
import pandas as pd
from pandasai import SmartDataframe
```

```python
df = pd.read_csv("teen_phone_addiction_dataset.csv")
df.head()
```

```python
sdf = SmartDataframe(df)
```

```python
try:
    result = sdf.chat("What is the average Daily_Usage_Hours?")
    print(result)
except Exception as e:
    print("PandasAI Error:")
    print(e)
```

```python
print("1. Average Daily Usage Hours:")
print(df["Daily_Usage_Hours"].mean())

print("\n2. Number of Male Students:")
print(df[df["Gender"] == "Male"].shape[0])

print("\n3. Most Common Addiction Level:")
print(df["Addiction_Level"].mode()[0])
```

```python
comparison = pd.DataFrame({
    "Question": [
        "Average Daily Usage Hours",
        "Number of Male Students",
        "Most Common Addiction Level"
    ],
    "PandasAI Result": [
        "Failed (API Key Required)",
        "Failed (API Key Required)",
        "Failed (API Key Required)"
    ],
    "Manual Pandas Result": [
        df["Daily_Usage_Hours"].mean(),
        df[df["Gender"] == "Male"].shape[0],
        df["Addiction_Level"].mode()[0]
    ]
})

comparison
```

```python
queries = [
    "What is the average Daily_Usage_Hours?",
    "How many male students are there?",
    "How many female students are there?",
    "What is the average Sleep_Hours?",
    "What is the most common Addiction_Level?",
    "What is the maximum Academic_Performance?",
    "What is the minimum Anxiety_Level?",
    "How many students use phones for Education?",
    "What is the average Time_on_Social_Media?",
    "Show the distribution of Gender."
]

print("10 Natural Language Queries:")
for i, query in enumerate(queries, start=1):
    print(f"{i}. {query}")
```

```python
error_log = []

for query in queries:
    try:
        result = sdf.chat(query)
        print(f"Query: {query}")
        print(result)
        print("-" * 50)
    except Exception as e:
        error_log.append({
            "Query": query,
            "Error": str(e)
        })

print("Completed!")
```

```python
error_df = pd.DataFrame(error_log)
error_df
```
