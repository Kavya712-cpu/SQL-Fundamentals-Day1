```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("teen_phone_addiction_dataset.csv")

df.head()
```

# Business Question

How does daily mobile phone usage affect students' academic performance and what recommendations can help improve academic outcomes?

```python
print(df.isnull().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print(df.describe())

print(df["Gender"].value_counts())

print(df["Academic_Performance"].describe())
```

```python
plt.figure(figsize=(6,4))

plt.scatter(df["Daily_Usage_Hours"], df["Academic_Performance"])

plt.title("Daily Phone Usage vs Academic Performance")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Academic Performance")

plt.show()
```

```python
correlation = df["Daily_Usage_Hours"].corr(df["Academic_Performance"])

print("Correlation between Daily Usage Hours and Academic Performance:")
print(correlation)
```
