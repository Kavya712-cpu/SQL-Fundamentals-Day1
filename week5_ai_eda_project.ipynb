```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("teen_phone_addiction_dataset.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
print(df.describe())
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Daily_Usage_Hours"], bins=10)
plt.title("Distribution of Daily Phone Usage")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Number of Students")
plt.show()
```

```python
avg_usage = df.groupby("Gender")["Daily_Usage_Hours"].mean()

plt.figure(figsize=(6,4))
avg_usage.plot(kind="bar")
plt.title("Average Daily Phone Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Daily Usage Hours")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(
    df["Daily_Usage_Hours"],
    df["Academic_Performance"]
)
plt.title("Daily Phone Usage vs Academic Performance")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Academic Performance")
plt.show()
```
