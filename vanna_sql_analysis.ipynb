```python
import sqlite3
import pandas as pd

df = pd.read_csv("teen_phone_addiction_dataset.csv")

conn = sqlite3.connect("student_data.db")

df.to_sql("students", conn, if_exists="replace", index=False)

print("Database created successfully!")
```

```python
query = """
SELECT COUNT(*) AS Total_Students
FROM students;
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT AVG(Daily_Usage_Hours) AS Average_Usage
FROM students;
"""

print(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Gender,
AVG(Daily_Usage_Hours) AS Average_Usage
FROM students
GROUP BY Gender;
"""

print(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT *
FROM students
ORDER BY Daily_Usage_Hours DESC
LIMIT 10;
"""

print(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT AVG(Academic_Performance) AS Average_Academic
FROM students;
"""

print(pd.read_sql_query(query, conn))
```

```python
vn.connect_to_sqlite("student_data.db")
```

```python
vn.train(ddl="""
CREATE TABLE students (
    Student_ID INTEGER,
    Gender TEXT,
    Daily_Usage_Hours REAL,
    Academic_Performance REAL
);
""")
```

```python
vn.train(
    question="How many students are there?",
    sql="SELECT COUNT(*) FROM students;"
)
```

```python
vn.train(
    question="What is the average daily phone usage?",
    sql="SELECT AVG(Daily_Usage_Hours) FROM students;"
)
```

```python
vn.train(
    question="Show average usage by gender.",
    sql="""
    SELECT Gender,
    AVG(Daily_Usage_Hours)
    FROM students
    GROUP BY Gender;
    """
)
```

```python
vn.train(
    question="Show top 10 phone users.",
    sql="""
    SELECT *
    FROM students
    ORDER BY Daily_Usage_Hours DESC
    LIMIT 10;
    """
)
```

```python
vn.train(
    question="What is the average academic performance?",
    sql="SELECT AVG(Academic_Performance) FROM students;"
)
```
